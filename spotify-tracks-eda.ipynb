```python
df["duration_min"] = df["duration_ms"] / 60000
```

```python
df["popularity_group"] = pd.cut(
    df["popularity"],
    bins=[-1, 20, 40, 60, 80, 100],
    labels=["Very Low", "Low", "Medium", "High", "Very High"]
)
```

```python
energy_bins = [-0.01, 0.3, 0.6, 0.8, 1]
energy_labels = ["Low", "Medium", "High", "Very High"]

df["energy_level"] = pd.cut(
    df["energy"],
    bins=energy_bins,
    labels=energy_labels
)

df["energy_level"].value_counts().sort_index()
```

```python
dance_bins = [-0.01, 0.3, 0.6, 0.8, 1]
dance_labels = ["Low", "Medium", "High", "Very High"]

df["danceability_level"] = pd.cut(
    df["danceability"],
    bins=dance_bins,
    labels=dance_labels
)

df["danceability_level"].value_counts().sort_index()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(df["popularity"], bins=30, color="#1DB954")

plt.title("Distribution of Track Popularity")
plt.xlabel("Popularity Score (0-100)")
plt.ylabel("Number of Tracks")

plt.show()
```

**Insight:** Popularity is heavily right-skewed — over 15,800 tracks (~14%) score 0, while very high scores (80+) are rare. Most tracks never break through.

```python
plt.figure(figsize=(8, 5))

order = ["Very Low", "Low", "Medium", "High", "Very High"]
sns.countplot(data=df, x="popularity_group", order=order, color="#1DB954")

plt.title("Number of Tracks by Popularity Group")
plt.xlabel("Popularity Group")
plt.ylabel("Number of Tracks")

plt.show()
```

**Insight:** Only 953 tracks (<1%) reach "Very High" popularity, versus ~66,900 in the two lowest groups combined — true hits are rare relative to catalog size.

```python
top_genres = (
    df.groupby("track_genre")["popularity"]
    .mean()
    .sort_values(ascending=False)
    .head(15)
)

plt.figure(figsize=(9, 7))
sns.barplot(x=top_genres.values, y=top_genres.index, color="#1DB954")

plt.title("Top 15 Genres by Average Popularity")
plt.xlabel("Average Popularity")
plt.ylabel("Genre")

plt.show()
```

**Insight:** Genre alone drives large popularity swings: "pop-film" tops out at 59.3 while "iranian", "romance", and "latin" average under 9 — a ~57-point gap.

```python
plt.figure(figsize=(8, 5))

order_e = ["Low", "Medium", "High", "Very High"]
sns.countplot(data=df, x="energy_level", order=order_e, color="#191414")

plt.title("Track Distribution by Energy Level")
plt.xlabel("Energy Level")
plt.ylabel("Number of Tracks")

plt.show()
```

**Insight:** Over 60% of tracks fall in High/Very High energy, reflecting the dataset's heavy mix of dance, electronic, and metal genres.

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="danceability_level", order=order_e, color="#1DB954")

plt.title("Track Distribution by Danceability Level")
plt.xlabel("Danceability Level")
plt.ylabel("Number of Tracks")

plt.show()
```

**Insight:** About 84% of tracks sit in the Medium-High danceability range, typical of a commercial, dance-friendly catalog.

```python
plt.figure(figsize=(7, 5))

sns.boxplot(data=df, x="explicit", y="popularity", color="#1DB954")

plt.title("Popularity by Explicit Content Flag")
plt.xlabel("Explicit")
plt.ylabel("Popularity")

plt.show()
```

**Insight:** Explicit tracks (only 8.6% of the data) average 36.5 popularity vs. 33.0 for non-explicit — a small but consistent edge, likely genre-driven (hip-hop, contemporary pop).

```python
num_cols = [
    "popularity", "danceability", "energy", "loudness", "speechiness",
    "acousticness", "instrumentalness", "liveness", "valence", "tempo", "duration_min"
]

plt.figure(figsize=(10, 8))
sns.heatmap(df[num_cols].corr(), annot=True, fmt=".2f", cmap="Greens")

plt.title("Correlation Between Audio Features")

plt.show()
```

**Insight:** No single audio feature correlates strongly with popularity (all under |0.1|) — but features correlate strongly with each other, e.g. energy-loudness (+0.76) and energy-acousticness (-0.73).

```python
sample = df.sample(4000, random_state=42)

plt.figure(figsize=(8, 6))
sns.scatterplot(data=sample, x="acousticness", y="energy", alpha=0.35, color="#191414", s=15)

plt.title("Energy vs. Acousticness")
plt.xlabel("Acousticness")
plt.ylabel("Energy")

plt.show()
```

**Insight:** A clear inverse relationship (-0.73): acoustic tracks tend to have lower energy, while heavily produced/electronic tracks run higher energy.

```python
plt.figure(figsize=(9, 5))

sns.histplot(df[df["duration_min"] < 10]["duration_min"], bins=40, color="#1DB954")

plt.title("Distribution of Track Duration (Minutes, <10 min)")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Tracks")

plt.show()
```

**Insight:** Average duration is ~3.8 minutes, with the middle 50% between 2.9 and 4.4 minutes — a typical commercial pop range. 598 tracks exceed 10 minutes and 838 run under 1 minute.

```python
plt.figure(figsize=(9, 5))

sns.histplot(df["tempo"], bins=40, color="#1DB954")

plt.title("Distribution of Track Tempo")
plt.xlabel("Tempo (BPM)")
plt.ylabel("Number of Tracks")

plt.show()
```

**Insight:** Tempo centers around 122 BPM (median 122.0), with the middle 50% between 99 and 140 BPM — consistent with typical pop/dance production; 157 tracks show tempo=0, likely a detection artifact.

```python
plt.figure(figsize=(9, 5))

sns.histplot(df["valence"], bins=30, color="#191414")

plt.title("Distribution of Valence (Musical Positivity)")
plt.xlabel("Valence (0 = Sad/Negative, 1 = Happy/Positive)")
plt.ylabel("Number of Tracks")

plt.show()
```

**Insight:** Valence is fairly evenly spread (mean 0.47, median 0.46) with no strong skew — the catalog mixes upbeat and melancholic tracks in roughly equal measure.

```python
top_artists = df["artists"].value_counts().head(10)

plt.figure(figsize=(9, 6))
sns.barplot(x=top_artists.values, y=top_artists.index, color="#191414")

plt.title("Top 10 Artists by Number of Tracks")
plt.xlabel("Number of Tracks")
plt.ylabel("Artist")

plt.show()
```

**Insight:** The Beatles lead with 279 rows, followed by George Jones and Stevie Wonder. This reflects the same track appearing under multiple genres (89,741 unique track_ids out of 113,550 rows), not raw output volume.

```python
mode_counts = df["mode"].value_counts().sort_index()
labels = ["Minor (0)", "Major (1)"]

plt.figure(figsize=(6, 5))
plt.pie(mode_counts.values, labels=labels, autopct="%1.1f%%", colors=["#191414", "#1DB954"], startangle=90)

plt.title("Track Distribution by Musical Mode")

plt.show()
```

**Insight:** 63.8% of tracks are in a Major key vs. 36.2% Minor, consistent with commercial music's general lean toward the brighter, major-key sound.

# Conclusion

Across 113,550 Spotify tracks (post-cleaning), popularity is **highly skewed and long-tail**: most tracks score low or zero, and fewer than 1% reach "Very High" popularity.

**Key findings:**
- **Genre is the strongest signal tied to popularity** — a ~57-point gap separates the top genre (pop-film) from the bottom (iranian).
- **Audio features alone barely correlate with popularity** (all |r| < 0.1) — factors outside the audio itself (artist, marketing, timing) likely matter more.
- **Audio features correlate logically with each other**: energy tracks closely with loudness (+0.76) and inversely with acousticness (-0.73).
- **The catalog skews "commercial"**: ~3.8 min average duration, mostly Medium-High danceability, and a Major-key majority.
- **Explicit tracks**, though only 8.6% of the data, average slightly higher popularity — likely a genre effect (hip-hop, contemporary pop).

**Takeaway:** audio features alone are not enough to predict popularity — genre, artist reputation, and external signals (streams, release timing, marketing) would need to be incorporated for a stronger model.
